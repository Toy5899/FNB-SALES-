```sql
-- %sql
--- checking what is in my table
SELECT * 
FROM fnbcasestudy.default.sales_case_study_raw_data;

SELECT COUNT(*) --checking how many rows
FROM fnbcasestudy.default.sales_case_study_raw_data;

SELECT * ---checking for nulls 
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE Date IS NULL
   OR Sales IS NULL
   OR `Cost Of Sales` IS NULL
   OR `Quantity Sold` IS NULL;

SELECT Date,COUNT(*) AS date_count --- checking for duplicate on the date column
FROM  fnbcasestudy.default.sales_case_study_raw_data
GROUP BY Date
HAVING COUNT(*) > 1
ORDER BY Date;

SELECT * ---checking if there are0 or negative quantities on the column quantity sold
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` <= 0;


SELECT * ----checking if thereare negative sales or cost of sales
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE Sales < 0
OR `Cost Of Sales` < 0;


SELECT ---Checking for the total sales, total cost of sales, total profit and total quantity sold
      SUM(Sales) AS Total_Sales,-- Adding all sales
      SUM(`Cost of Sales`) AS Total_Cost,       -- Adding all costs
      SUM(Sales - `Cost of Sales`) AS Total_Profit, -- Sales minus cost of sales
      SUM(`Quantity Sold`) AS Total_Quantity     -- Adding all units sold
FROM fnbcasestudy.default.sales_case_study_raw_data;



SELECT --calculating the gross profit percentage
    SUM(Sales) AS Total_Sales,  --- total sales                    
    SUM(Sales - `Cost of Sales`) AS Total_Profit, -- Total profit
    (SUM(Sales - `Cost of Sales`) / SUM(Sales)) * 100 --calculating the gross profit percentage
    AS Gross_Profit_Percentage -- Gross profit percentage
FROM fnbcasestudy.default.sales_case_study_raw_data;


SELECT *, --- creating day of week classification
    CASE
        WHEN DAYOFWEEK(Date) = 2 THEN 'Monday'
        WHEN DAYOFWEEK(Date) = 3 THEN 'Tuesday'
        WHEN DAYOFWEEK(Date) = 4 THEN 'Wednesday'
        WHEN DAYOFWEEK(Date) = 5 THEN 'Thursday'
        WHEN DAYOFWEEK(Date) = 6 THEN 'Friday'
        WHEN DAYOFWEEK(Date) = 7 THEN 'Saturday'
        WHEN DAYOFWEEK(Date) = 1 THEN 'Sunday'
    END AS Day_Classification
FROM fnbcasestudy.default.sales_case_study_raw_data;

SELECT *, -- creating day classification
    CASE
        WHEN DAYOFWEEK(Date) IN (1, 7) THEN 'Weekend'
        ELSE 'Weekday'
    END AS Day_Classification
FROM fnbcasestudy.default.sales_case_study_raw_data;


SELECT ---checking the day with highest sales
    Date, 
    Sales,       
    `Quantity Sold`          
FROM fnbcasestudy.default.sales_case_study_raw_data
ORDER BY Sales DESC
LIMIT 1;

SELECT ---checking the day with lowest sales
    Date,                  
    Sales,                
    `Quantity Sold`        
FROM fnbcasestudy.default.sales_case_study_raw_data
ORDER BY Sales ASC
LIMIT 1;



CREATE OR REPLACE TEMP VIEW sales_case_study AS ---creating a temporary table

SELECT
    Date,
    Sales,
    `Cost of Sales`,
    `Quantity Sold`,

    Sales / `Quantity Sold` AS Sales_Price_Per_Unit,

    Sales - `Cost of Sales` AS Gross_Profit,

    ((Sales - `Cost of Sales`) / Sales) * 100 AS Gross_Profit_Percentage,

    (Sales - `Cost of Sales`) / `Quantity Sold`
        AS Gross_Profit_Per_Unit,

    CASE
        WHEN DAYOFWEEK(Date) = 2 THEN 'Monday'
        WHEN DAYOFWEEK(Date) = 3 THEN 'Tuesday'
        WHEN DAYOFWEEK(Date) = 4 THEN 'Wednesday'
        WHEN DAYOFWEEK(Date) = 5 THEN 'Thursday'
        WHEN DAYOFWEEK(Date) = 6 THEN 'Friday'
        WHEN DAYOFWEEK(Date) = 7 THEN 'Saturday'
        WHEN DAYOFWEEK(Date) = 1 THEN 'Sunday'
    END AS Day_Classification

FROM fnbcasestudy.default.sales_case_study_raw_data;


SELECT --- checking the selling price per unit for each day
    Date,
    Sales,
    `Quantity Sold`,
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit ---calculating the selling price for one unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
ORDER BY Sales_Price_Per_Unit ASC;

SELECT ---calculating performance during promotion period
    MIN(Sales / `Quantity Sold`) AS Lowest_Price,
    MAX(Sales / `Quantity Sold`) AS Highest_Price,
    AVG(Sales / `Quantity Sold`) AS Average_Price

FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0;

SELECT --checking the days with the lowest selling prices
    Date,
    Sales,
    `Quantity Sold`,
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
ORDER BY Sales_Price_Per_Unit ASC
LIMIT 20;


SELECT --checking sales around a promotion period
    Date,
    Sales,
    `Quantity Sold`,
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
ORDER BY Date;

SELECT -- checking the next lowest prices after the first promotion period
    Date,                                    
    Sales,                                     
    `Quantity Sold`,                         
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
AND Date NOT BETWEEN '2013-12-30' AND '2014-01-01'
ORDER BY Sales_Price_Per_Unit ASC
LIMIT 20;


SELECT-- checking another low period
    Date,                                     
    Sales,                                     
    `Quantity Sold`,                           
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
AND Date NOT BETWEEN '2013-12-30' AND '2014-01-01'
AND Date NOT BETWEEN '2014-09-01' AND '2014-09-02'
ORDER BY Sales_Price_Per_Unit ASC
LIMIT 20;


SELECT ---checking price during september 2014
    Date,                                    
    Sales,                                  
    `Quantity Sold`,                           
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE Date BETWEEN '2014-08-25' AND '2014-09-15'
AND `Quantity Sold` > 0
ORDER BY Date;


SELECT ---checking all the price from 25 august to 15 september
    Date,
    Sales,
    `Quantity Sold`,
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE Date BETWEEN '2014-08-25' AND '2014-09-15'
AND `Quantity Sold` > 0
ORDER BY Date;



SELECT -- checking another low price period
    Date,
    Sales,
    `Quantity Sold`,
    Sales / `Quantity Sold` AS Sales_Price_Per_Unit
FROM fnbcasestudy.default.sales_case_study_raw_data
WHERE `Quantity Sold` > 0
AND Date NOT BETWEEN '2013-12-30' AND '2014-01-01'
AND Date NOT BETWEEN '2014-08-28' AND '2014-09-08'
ORDER BY Sales_Price_Per_Unit ASC
LIMIT 30;


```
